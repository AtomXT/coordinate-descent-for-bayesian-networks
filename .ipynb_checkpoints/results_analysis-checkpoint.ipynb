# obj difference

```python
cd_results_obj = pd.read_csv('./experiment results/synthetic_results_CD_order_small_diff.csv')
ges_results_obj = pd.read_csv('./experiment results/synthetic_results_ges_small_alpha.csv')
micodag_results_obj = pd.read_csv('./experiment results/synthetic_results_micodag_small_diff.csv')
cd_results_obj['Method'] = 'Our method'
micodag_results_obj['Method'] = 'MICODAG'
ges_results_obj['Method'] = 'GES'
micodag_results_obj = micodag_results_obj[micodag_results_obj['RGAP'] < 0.01]
all_results_obj1 = pd.concat([cd_results_obj, micodag_results_obj, ges_results_obj])
cd_obj_diff_ratio = (all_results_obj1[all_results_obj1.Method == 'Our method'].obj - all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj)/all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj
cd_obj_diff_ratio[cd_obj_diff_ratio<0] = 0
ges_obj_diff_ratio = (all_results_obj1[all_results_obj1.Method == 'GES'].obj - all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj)/all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj
ges_obj_diff_ratio[ges_obj_diff_ratio<0] = 0
obj_diff_ratio = pd.DataFrame(micodag_results_obj[['graph', 'n_sample', 'iter']])
obj_diff_ratio['Our method'] = cd_obj_diff_ratio
obj_diff_ratio['GES'] = ges_obj_diff_ratio
obj_diff_ratio = obj_diff_ratio.set_index(['graph', 'n_sample'])
```

```python
obj_diff_mean = obj_diff_ratio['Our method'].groupby(level=['n_sample','graph']).mean()
temp_obj = obj_diff_mean.unstack(['graph'])
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], temp_obj.iloc[1:,0], "-x")
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], temp_obj.iloc[1:,1], '-o')
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], temp_obj.iloc[1:,2], '-s')
plt.xlabel('n', fontsize=15)
plt.ylabel('Percentage of differences', fontsize=15)
# plt.legend(['Graph1 (7 edges)', 'Graph2 (12 edges)'], fontsize=15)
plt.legend(['Graph1 (7 edges)', 'Graph2 (12 edges)', 'Graph3 (21 edges)'], fontsize=15)
plt.savefig('./experiment results/plots/obj_diff_new.pdf', bbox_inches='tight')
plt.show()
```

```python
ges_obj_diff_mean = obj_diff_ratio['GES'].groupby(level=['n_sample', 'graph']).mean()
ges_temp_obj = ges_obj_diff_mean.unstack(['graph'])
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], ges_temp_obj.iloc[1:,0], "-x")
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], ges_temp_obj.iloc[1:,1], '-o')
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], ges_temp_obj.iloc[1:,2], '-s')
plt.xlabel('n', fontsize=15)
plt.ylabel('Percentage of differences', fontsize=15)
plt.legend(['Graph1 (7 edges)', 'Graph2 (12 edges)', 'Graph3 (21 edges)'], fontsize=15)
# plt.legend(['Graph1 (7 edges)', 'Graph2 (12 edges)'], fontsize=15)
plt.savefig('./experiment results/plots/ges_obj_diff_new.pdf', bbox_inches='tight')
plt.show()
```

```python
all_results_obj = pd.concat([cd_results_obj, micodag_results_obj, ges_results_obj])
all_results_obj = all_results_obj.set_index(['graph', 'n_sample', 'Method'])
obj_time_mean = all_results_obj['time'].groupby(level=['n_sample','graph', 'Method']).mean()
temp = obj_time_mean.unstack(['graph', 'Method'])
obj_time_figure = plt.figure()
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200],temp.iloc[1:,8],'-x', label='CD', color='red')
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200],temp.iloc[1:,7], '-s', label='MICODAG', color='c')
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200],temp.iloc[1:,6],'--o', label='GES', color='blue')
plt.legend(fontsize=15)
plt.grid(False)
plt.ylabel('Time (seconds)', fontsize=15)
plt.xlabel('n', fontsize=15)
plt.savefig('./experiment results/plots/obj_time_graph3_new.pdf', bbox_inches='tight')
plt.show()
```

# Objective convergence with random orderings

```python
cd_results0_obj = pd.read_csv('./experiment results/synthetic_results_CD_order_small_diff_random_ordering0.csv')
cd_results1_obj = pd.read_csv('./experiment results/synthetic_results_CD_order_small_diff_random_ordering1.csv')
cd_results2_obj = pd.read_csv('./experiment results/synthetic_results_CD_order_small_diff_random_ordering2.csv')
micodag_results_obj = pd.read_csv('./experiment results/synthetic_results_micodag_small_diff.csv')
cd_results0_obj['Method'] = 'CD random ordering1'
cd_results1_obj['Method'] = 'CD random ordering2'
cd_results2_obj['Method'] = 'CD random ordering3'
micodag_results_obj['Method'] = 'MICODAG'

micodag_results_obj = micodag_results_obj[micodag_results_obj['RGAP'] < 0.01]
all_results_obj1 = pd.concat([cd_results0_obj, cd_results1_obj,cd_results2_obj, micodag_results_obj])
cd_obj_diff_ratio1 = (all_results_obj1[all_results_obj1.Method == 'CD random ordering1'].obj - all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj)/all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj
cd_obj_diff_ratio1[cd_obj_diff_ratio1<0] = 0
cd_obj_diff_ratio2 = (all_results_obj1[all_results_obj1.Method == 'CD random ordering2'].obj - all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj)/all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj
cd_obj_diff_ratio2[cd_obj_diff_ratio2<0] = 0
cd_obj_diff_ratio3 = (all_results_obj1[all_results_obj1.Method == 'CD random ordering3'].obj - all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj)/all_results_obj1[all_results_obj1.Method == 'MICODAG'].obj
cd_obj_diff_ratio3[cd_obj_diff_ratio3<0] = 0

obj_diff_ratio = pd.DataFrame(micodag_results_obj[['graph', 'n_sample', 'iter']])
obj_diff_ratio['CD(ordering1)'] = cd_obj_diff_ratio1
obj_diff_ratio['CD(ordering2)'] = cd_obj_diff_ratio2
obj_diff_ratio['CD(ordering3)'] = cd_obj_diff_ratio3
obj_diff_ratio = obj_diff_ratio.loc[obj_diff_ratio['graph'] == 1,]
obj_diff_ratio = obj_diff_ratio.set_index(['n_sample'])
```

```python
obj_diff_ratio.groupby(level=['n_sample']).mean()
```

```python
obj_diff_mean = obj_diff_ratio.groupby(level=['n_sample']).mean()
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], obj_diff_mean.iloc[1:,2], "-x")
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], obj_diff_mean.iloc[1:,3], '-o')
plt.plot([100, 200, 300, 400, 500, 800, 1000, 1600, 3200], obj_diff_mean.iloc[1:,4], '-s')
plt.xlabel('n', fontsize=15)
plt.ylabel('Percentage of differences', fontsize=15)
# plt.legend(['Graph1 (7 edges)', 'Graph2 (12 edges)'], fontsize=15)
plt.legend(['CD (random ordering1)', 'CD (random ordering2)', 'CD (random ordering3)'], fontsize=15)
plt.savefig('./experiment results/plots/obj_diff_random_ordering.pdf', bbox_inches='tight')
plt.show()
```

```python
# Using the graph to choose v-strcutrue to get a controlled setting of incorrect ordering.
import networkx as nx
import matplotlib.pyplot as plt 
import os


datasets = ['1dsep', '2asia', '3bowling', '4insuranceSmall', '5rain', '6cloud', '7funnel', '8galaxy', '9insurance', '10factors', '11hfinder', '12hepar']
```

```python
G = nx.DiGraph()
network = '12hepar'
file_path = "./Data/RealWorldDatasetsTXu_smallalpha1/" + f"{network}"
graph_name = [i for i in os.listdir(
        file_path) if os.path.isfile(os.path.join(file_path, i)) and 'Sparse_Original_edges' in i][0]
graph_path = file_path + f"/{graph_name}"
df = pd.read_table(graph_path, delimiter=',', dtype=int, header=None)
edges = list(zip(df.iloc[:,0], df.iloc[:,1]))
G.add_edges_from(edges)

plt.figure(figsize=(8,8))
nx.draw(G,with_labels=True)
```
